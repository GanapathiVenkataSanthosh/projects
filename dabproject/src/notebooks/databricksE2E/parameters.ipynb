```python
df = spark.read.option("multiline", "true").json("abfss://source@databricksgvse2e.dfs.core.windows.net/FileNames.json")

from pyspark.sql.functions import explode

df_exploded = df.select(explode(df["dataset"]).alias("files"))

filenames_df = df_exploded.select("files.filename")

filenames = [row.filename for row in filenames_df.collect()]
dbutils.jobs.taskValues.set("filenames", filenames)
```
