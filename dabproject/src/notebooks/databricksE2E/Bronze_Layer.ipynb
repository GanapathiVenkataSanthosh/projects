## # Reading Data using Autoloader

```python
df1 = spark.readStream.format("cloudFiles") \
  .option("cloudFiles.format","parquet") \
  .option("cloudFiles.schemaLocation",f"abfss://bronze@databricksgvse2e.dfs.core.windows.net/schema/{p_file_name}_schema")\
  .load(f"abfss://source@databricksgvse2e.dfs.core.windows.net/{p_file_name}")
```

## Writing data using Autoloader

```python
df1.writeStream.format("parquet")\
    .option("mode", "update")\
    .option("checkpointLocation",f"abfss://bronze@databricksgvse2e.dfs.core.windows.net/schema/{p_file_name}_schema")\
    .option("path", f"abfss://bronze@databricksgvse2e.dfs.core.windows.net/{p_file_name}")\
    .trigger(once=True)\
    .start()
```
